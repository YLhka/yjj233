```python
# 5. 编写一个Python函数，判断一个整数是否是回文数。回文数是指正序（从左向右）和倒序（从右向左）读都是一样的数。
def is_palindrome(num):
    return str(num)==str(num)[::-1]

number=input("请输入一个整数:")
print(is_palindrome(number))
```

```python
# 6.编写一个Python函数，返回一个包含斐波那契数列的前N个数的列表。斐波那契数列是一个数列，其中每个数是前两个数的和，数列以0和1开始。
def fibonacci(n):
    if n<=0: fib_list=[]

    elif n==1: fib_list=[0]

    else: 
        fib_list=[0,1]
        for i in range(2,n):
            fib_list.append(fib_list[i-1]+fib_list[i-2])
    return fib_list

n=eval(input("请输入一个整数："))
print(fibonacci(n))

        
```

```python
# 7.编写一个Python函数，接受一个整数数组作为输入，返回数组中第二大的数。如果数组中只有一个元素或数组为空，则返回-1。
def second_largest(nums):
    if len(nums) < 2:
        return -1

    # 初始化 largest 和 second 为负无穷大
    largest = float('-inf')
    second = float('-inf')
    
    for num in nums:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num

    # 如果 second 仍然是负无穷大，说明没有找到第二大的数
    if second == float('-inf'):
        return -1
    else:
        return second
        
# 获取用户输入
input_string = input("请输入一个整数数组，用逗号分隔: ")
# 将输入的字符串转换为整数列表
nums = list(map(int, input_string.split(',')))

# 调用函数并输出结果
result = second_largest(nums)
print(f"数组中的第二大的数是: {result}")
```

```python
# 8. 编写一个函数，使其能够用正则表达式的方式简单验证身份证号是否合法。
import re

def validate_id_card(id_card):
    # 定义正则表达式模式
    pattern = r'^[1-9]\d{5}(19|20)\d{2}(0[1-9]|1[0-2])(0[1-9]|[12]\d|3[01])\d{3}(\d|X|x)$'
    
    # 使用正则表达式匹配身份证号
    if re.match(pattern, id_card):
        return True
    else:
        return False

user_id = input("请输入身份证号：")
if validate_id_card(user_id):
    print("身份证号合法！")
else:
    print("身份证号不合法！")
```
