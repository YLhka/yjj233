```python
# 写一个 Python 程序，判断一个输入的字符串 S 是否包含由两个或两个以上连续出现的相同字符组成的字符串。例如，abccccda 中就包含 cccc 这个由 4 个连续字符组成的字符串。
s=input('请输入字符串:')
a='0'
for i in s:
    if i==a:
        print('包含')
        break
    else:
        a=i
else:
    print('不包含')
```

```python
# 请设计一个求 3 次方根的算法（不允许直接调用求方根的函数），并给出对应的 Python 程序。 采用牛顿迭代法求解，迭代过程数学公式表示为： 
def cube_root(a, epsilon=1e-7, max_iter=1000):
    """
    使用牛顿迭代法计算a的三次方根。
    
    参数:
    a -- 要求立方根的数
    epsilon -- 迭代停止的精度，默认为1e-7
    max_iter -- 最大迭代次数，默认为1000
    
    返回:
    x -- a的三次方根的近似值
    """
    # 初始猜测值可以是a本身，或者a的某个比例，这里为了通用性直接取a
    x_n = a
    
    for _ in range(max_iter):
        # 应用牛顿迭代公式
        x_n_next = x_n - (x_n**3 - a) / (3 * x_n**2)
        
        # 检查收敛性
        if abs(x_n_next - x_n) < epsilon:
            return x_n_next
        
        x_n = x_n_next
    
    # 如果达到最大迭代次数仍未满足精度要求，返回最后一次迭代的结果
    print("警告: 达到最大迭代次数，结果可能不精确。")
    return x_n

# 示例：计算27的三次方根
number = eval(input("请输入要计算的数字: "))
result = cube_root(number)
print(number,"的三次方根约为:" ,result)
```

```python
# 写一个 Python 程序，给定一个常数 n（n>0），求 n 的阶乘，即 n!=1×2×...×(n-1)×n。例如，4!=24，5!=120。
s=1
n=eval(input("请输入一个正整数："))
for i in range(1,n+1):
    s*=i
print(s)
```
